```python
import numpy as np
```

```python
df_playstore = df_playstore[df_playstore['Rating'] <= 5]


df_playstore ['Installs'] = pd.to_numeric(df_playstore['Installs'].astype(str).str.replace(r'[+,]', '', regex=True), errors='coerce')
df_playstore ['Price'] = pd.to_numeric(df_playstore['Price'].astype(str).str.replace(r'[$,]', '', regex=True), errors='coerce')
df_playstore ['Reviews'] = pd.to_numeric(df_playstore['Reviews'], errors='coerce')
```

```python
def clean_size(val):
    val = str(val).replace(',','')
    if 'M' in val:
        return float(val.replace('M',''))
    elif 'k' in val:
        return float(val.replace('k',''))/1024
    return np.nan
```

```python
df_playstore['Size_MB'] = df_playstore['Size'].apply(clean_size)
```

```python
df_playstore_clean = pd.DataFrame({
    'App': df_playstore['App'],
    'Platform': 'Google Play Store',
    'Category': df_playstore['Category'],
    'Rating': df_playstore['Rating'],
    'Reviews': df_playstore['Reviews'],
    'Size_MB': df_playstore['Size_MB'],
    'Installs': df_playstore['Installs'],
    'Type': df_playstore['Type'],
    'Price': df_playstore['Price'],
    'Content Rating': df_playstore['Content Rating'],
    'Genres': df_playstore['Genres'],
    'Last Updated': pd.to_datetime(df_playstore['Last Updated'], errors='coerce'),
    'Current Ver': df_playstore['Current Ver'],
    'Android Ver': df_playstore['Android Ver']
})
```

```python
df_appstore['size_bytes'] = df_appstore['size_bytes'] / 1024 / 1024
```

```python
df_appstore_clean = pd.DataFrame({
    'App': df_appstore['track_name'],
    'Platform': 'Apple App Store',
    'Category': df_appstore['prime_genre'],
    'Rating': df_appstore['user_rating'],
    'Reviews': df_appstore['rating_count_tot'],
    'Size_MB': df_appstore['size_bytes'],
    'Price': df_appstore['price'],
}).dropna()
```

```python
df_combined = pd.concat([df_playstore_clean, df_appstore_clean], ignore_index=True)
```

```python
feature = df_combined[['Rating', 'Reviews', 'Price']]
for col in feature:
    df_combined[col] = pd.to_numeric(df_combined[col], errors='coerce')
    
```

```python
df_combined = df_combined.dropna(subset=feature.columns).reset_index(drop=True)
```

```python
X = df_combined[feature.columns].values
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df_combined['Cluster_ID'] = kmeans.fit_predict(x_scaled)
```

```python
df_combined.to_csv('combined_apps_with_clusters.csv', index=False)
print("Combined dataset with clusters saved to 'combined_apps_with_clusters.csv'")
```
